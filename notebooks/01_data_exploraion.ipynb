```python
s1_idx = s1.set_index("entity_id")
s2_idx = s2.set_index("entity_id")
s3_idx = s3.set_index("entity_id")
```

```python
def create_match_table(gt, s1_idx, s2_idx, s3_idx, n=20):

    rows = []

    for _, gt_row in gt.head(n).iterrows():

        s1_id = gt_row["source1_entity_id"]
        matched_ids = gt_row["matched_entity_ids"]

        # Source 1 record
        s1_record = s1_idx.loc[s1_id]

        # If there are no matches
        if pd.isna(matched_ids) or matched_ids == "":
            rows.append({
                "s1_id": s1_id,
                "s1_name": s1_record["business_name"],
                "s1_address": s1_record["business_address"],
                "s1_country": s1_record["country"],
                "matched_id": "",
                "matched_name": "",
                "matched_address": "",
                "matched_country": ""
            })
            continue

        # Multiple matches
        for matched_id in str(matched_ids).split(","):

            matched_id = matched_id.strip()

            if matched_id.startswith("S2-"):
                record = s2_idx.loc[matched_id]

            elif matched_id.startswith("S3-"):
                record = s3_idx.loc[matched_id]

            else:
                continue

            rows.append({
                "s1_id": s1_id,
                "s1_name": s1_record["business_name"],
                "s1_address": s1_record["business_address"],
                "s1_country": s1_record["country"],
                "matched_id": matched_id,
                "matched_name": record["business_name"],
                "matched_address": record["business_address"],
                "matched_country": record["country"]
            })

    return pd.DataFrame(rows)
```

```python
import re
import unicodedata
import pandas as pd


# ============================================================
# Generic helpers
# ============================================================

def normalize_unicode(text):
    """
    Normalize Unicode characters.

    Example:
        'Dréxkor' -> 'Dréxkor'
    """
    if pd.isna(text):
        return ""

    text = str(text)

    # Normalize Unicode representation
    text = unicodedata.normalize("NFKC", text)

    return text


def normalize_spaces(text):
    """Collapse multiple spaces into one."""
    return re.sub(r"\s+", " ", text).strip()


# ============================================================
# NAME NORMALIZATION
# ============================================================

# Conservative business abbreviation normalization.
# We are NOT removing these words.
NAME_ABBREVIATIONS = {
    "pvt": "private",
    "pvt.": "private",

    "ltd": "limited",
    "ltd.": "limited",

    "corp": "corporation",
    "corp.": "corporation",

    "co": "company",
    "co.": "company",

    "inc": "incorporated",
    "inc.": "incorporated",
}


def normalize_name(name):
    """
    Normalize a business name without attempting to resolve
    whether two names represent the same business.

    Operations:
    - missing values -> ""
    - Unicode normalization
    - lowercase
    - remove domain prefix
    - punctuation -> spaces
    - normalize selected abbreviations
    - normalize whitespace

    Important:
    We DO NOT perform typo correction.
    """

    if pd.isna(name):
        return ""

    name = str(name)

    # Unicode normalization
    name = unicodedata.normalize("NFKC", name)

    # Lowercase
    name = name.lower().strip()

    # Handle common domain-style business names
    # Example:
    # maurewilliamscolombier.com
    # -> maurewilliamscolombier
    name = re.sub(
        r"\.(com|net|org|in|co|biz|us)$",
        "",
        name
    )

    # Convert punctuation/symbols to spaces
    # Keep letters and numbers.
    name = re.sub(r"[^a-z0-9\s]", " ", name)

    # Normalize whitespace
    name = normalize_spaces(name)

    # Normalize known abbreviations token-by-token
    tokens = name.split()

    normalized_tokens = []

    for token in tokens:
        normalized_tokens.append(
            NAME_ABBREVIATIONS.get(token, token)
        )

    return " ".join(normalized_tokens)


# ============================================================
# ADDRESS NORMALIZATION
# ============================================================

ADDRESS_ABBREVIATIONS = {
    # Street types
    "rd": "road",
    "rd.": "road",

    "st": "street",
    "st.": "street",

    "ave": "avenue",
    "ave.": "avenue",

    "av": "avenue",
    "av.": "avenue",

    "blvd": "boulevard",
    "blvd.": "boulevard",

    "dr": "drive",
    "dr.": "drive",

    "ln": "lane",
    "ln.": "lane",

    "hwy": "highway",
    "hwy.": "highway",

    "pkwy": "parkway",
    "pkwy.": "parkway",

    "ct": "court",
    "ct.": "court",

    # Apartment/unit
    "apt": "apartment",
    "apt.": "apartment",

    "ste": "suite",
    "ste.": "suite",
}


def normalize_address(address):
    """
    Normalize an address conservatively.

    Operations:
    - missing values -> ""
    - Unicode normalization
    - lowercase
    - punctuation -> spaces
    - common address abbreviations
    - whitespace normalization

    Numbers are preserved.
    """

    if pd.isna(address):
        return ""

    address = str(address)

    # Unicode normalization
    address = unicodedata.normalize("NFKC", address)

    # Lowercase
    address = address.lower().strip()

    # Punctuation -> spaces
    # This preserves numbers.
    address = re.sub(r"[^a-z0-9\s]", " ", address)

    # Normalize whitespace
    address = normalize_spaces(address)

    # Token-level abbreviation normalization
    tokens = address.split()

    normalized_tokens = []

    for token in tokens:
        normalized_tokens.append(
            ADDRESS_ABBREVIATIONS.get(token, token)
        )

    return " ".join(normalized_tokens)


# ============================================================
# COUNTRY NORMALIZATION
# ============================================================

def normalize_country(country):
    """
    Country is treated as an open-set string.

    We do NOT restrict countries to US/India because
    the test set can contain France and potentially
    other labels.
    """

    if pd.isna(country):
        return ""

    country = str(country)

    country = unicodedata.normalize("NFKC", country)

    country = country.lower()

    country = normalize_spaces(country)

    return country


# ============================================================
# COMPLETE RECORD NORMALIZATION
# ============================================================

def preprocess_dataframe(df):
    """
    Add normalized columns without modifying original columns.
    """

    result = df.copy()

    result["name_norm"] = (
        result["business_name"]
        .apply(normalize_name)
    )

    result["address_norm"] = (
        result["business_address"]
        .apply(normalize_address)
    )

    result["country_norm"] = (
        result["country"]
        .apply(normalize_country)
    )

    return result
```
